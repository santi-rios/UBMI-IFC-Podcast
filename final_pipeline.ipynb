## 3. Embeddings

Vamos a utilizar 2 métodos:
- **Gemini API**: Google's cloud embedding model
- **EmbeddingGemma**

```python
# Setup
device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"🔧 Usando dispositivo: {device}")

# EmbeddingGemma model
print("📥 Cargando modelo EmbeddingGemma...")
embeddinggemma_model_id = "google/embeddinggemma-300M"

try:
    # Load EmbeddingGemma using SentenceTransformers
    start_time = time.time()
    embeddinggemma_model = SentenceTransformer(embeddinggemma_model_id, device=device)
    load_time = time.time() - start_time
    
    print(f"✅ EmbeddingGemma cargado en {load_time:.1f} seconds")
    print(f"📊 Parámetros: {sum(p.numel() for p in embeddinggemma_model.parameters()):,}")
    
except Exception as e:
    print(f"❌ Error en EmbeddingGemma: {e}")
    print("💡 Note: You may need to accept the Gemma license and login to Hugging Face")
    print("Falling back to Sentence-BERT...")
    embeddinggemma_model = SentenceTransformer('all-MiniLM-L6-v2', device=device)
    print("✅ Using Sentence-BERT as fallback")

# Function to generate text for embedding (title + abstract)
def prepare_text_for_embedding(pub):
    """Combine title and abstract for embedding generation"""
    title = pub.get('title', '')
    abstract = pub.get('abstract', '')
    return f"{title}. {abstract}".strip()

# Updated rate limiting for Gemini embeddings generation
def generate_gemini_embeddings(texts, batch_size=50):
    """Generate embeddings using Gemini API with proper rate limiting"""
    print(f"🌐 Generando Gemini embeddings para {len(texts)} textos...")
    
    # Rate limits for Gemini Embedding (from your table)
    RPM_LIMIT = 100  # Requests per minute
    RPD_LIMIT = 1000  # Requests per day
    
    embeddings = []
    expected_dim = None
    failed_count = 0
    request_count = 0
    start_time = time.time()
    daily_request_count = 0
    
    # Calculate minimum interval between requests
    min_interval = 60.0 / RPM_LIMIT  # 0.6 seconds between requests
    last_request_time = 0
    
    for i in range(0, len(texts), batch_size):
        batch = texts[i:i+batch_size]
        batch_num = i//batch_size + 1
        total_batches = (len(texts)-1)//batch_size + 1
        
        print(f"🔄 Gemini batch {batch_num}/{total_batches} ({len(batch)} textos)")
        
        for j, text in enumerate(batch):
            # Check daily limit
            if daily_request_count >= RPD_LIMIT:
                print(f"⚠️ límite de la API ({RPD_LIMIT}). Parando por hoy...")
                break
                
            # Rate limiting: ensure we don't exceed RPM
            current_time = time.time()
            time_since_last = current_time - last_request_time
            
            if time_since_last < min_interval:
                sleep_time = min_interval - time_since_last
                time.sleep(sleep_time)
            
            try:
                # Ensure text is not too long for Gemini API
                text_trimmed = text[:8000] if len(text) > 8000 else text
                
                # Make API call with proper model name
                result = genai.embed_content(
                    model="models/gemini-embedding-001",
                    content=text_trimmed,
                    task_type="retrieval_document"
                )
                
                embedding = result['embedding']
                last_request_time = time.time()
                request_count += 1
                daily_request_count += 1
                
                # Check embedding dimension consistency
                if expected_dim is None:
                    expected_dim = len(embedding)
                    print(f"📏 dimensiones Gemini embedding: {expected_dim}")
                elif len(embedding) != expected_dim:
                    print(f"⚠️ Dimension mismatch: esperado {expected_dim}, obtenido {len(embedding)}")
                    # Pad or truncate to expected dimension
                    if len(embedding) < expected_dim:
                        embedding.extend([0.0] * (expected_dim - len(embedding)))
                    else:
                        embedding = embedding[:expected_dim]
                
                embeddings.append(embedding)
                
                # Progress update every 50 requests
                if request_count % 50 == 0:
                    elapsed = time.time() - start_time
                    rate = request_count / (elapsed / 60)  # requests per minute
                    print(f"   📈 Progreso: {request_count}/{len(texts)} requests ({rate:.1f} RPM, {daily_request_count}/{RPD_LIMIT} daily)")
                
            except Exception as e:
                error_msg = str(e)
                failed_count += 1
                
                if "quota" in error_msg.lower() or "429" in error_msg:
                    print(f"⚠️ Rate limit hit. Waiting 60 seconds...")
                    time.sleep(60)
                    # Retry once
                    try:
                        result = genai.embed_content(
                            model="models/gemini-embedding-001",
                            content=text_trimmed,
                            task_type="retrieval_document"
                        )
                        embedding = result['embedding']
                        last_request_time = time.time()
                        daily_request_count += 1
                        
                        if expected_dim and len(embedding) != expected_dim:
                            if len(embedding) < expected_dim:
                                embedding.extend([0.0] * (expected_dim - len(embedding)))
                            else:
                                embedding = embedding[:expected_dim]
                        
                        embeddings.append(embedding)
                        
                    except Exception as e2:
                        print(f"⚠️ Retry failed for text {i*batch_size + j + 1}: {str(e2)[:100]}")
                        fallback_dim = expected_dim if expected_dim else 768
                        embeddings.append([0.0] * fallback_dim)
                else:
                    print(f"⚠️ Gemini API error for text {i*batch_size + j + 1}: {str(e)[:100]}")
                    fallback_dim = expected_dim if expected_dim else 768
                    embeddings.append([0.0] * fallback_dim)
        
        # Check if we hit daily limit
        if daily_request_count >= RPD_LIMIT:
            break
    
    print(f"✅ Gemini embeddings listo!")
    print(f"📊 Generado: {len(embeddings)} embeddings")
    print(f"📊 Dinensiones: {len(embeddings[0]) if embeddings else 0}")
    print(f"⚠️ Failed/Fallback: {failed_count}")
    print(f"📊 Total API requests: {daily_request_count}/{RPD_LIMIT}")
    
    # Convert to numpy array
    try:
        embeddings_array = np.array(embeddings)
        print(f"📊 Final array shape: {embeddings_array.shape}")
        return embeddings_array
    except ValueError as e:
        print(f"❌ Array conversion error: {e}")
        print("🔧 Attempting to fix dimension inconsistencies...")
        
        if embeddings:
            max_dim = max(len(emb) for emb in embeddings)
            print(f"📏 Max dimension found: {max_dim}")
            
            fixed_embeddings = []
            for emb in embeddings:
                if len(emb) < max_dim:
                    emb_fixed = emb + [0.0] * (max_dim - len(emb))
                else:
                    emb_fixed = emb[:max_dim]
                fixed_embeddings.append(emb_fixed)
            
            embeddings_array = np.array(fixed_embeddings)
            print(f"✅ Fixed array shape: {embeddings_array.shape}")
            return embeddings_array
        
        raise

# Function to generate EmbeddingGemma embeddings
def generate_embeddinggemma_embeddings(texts, model, task_name="STS", batch_size=32):
    """Generate embeddings using EmbeddingGemma model"""
    print(f"🔄 Procesando {len(texts)} textos con EmbeddingGemma en batches de {batch_size}")
    
    # Track memory usage if on GPU
    if torch.cuda.is_available():
        torch.cuda.empty_cache()
        initial_memory = torch.cuda.memory_allocated() / 1e9
        print(f"💾 Memoria de GPU inicial: {initial_memory:.2f} GB")
    
    start_time = time.time()
    all_embeddings = []
    
    for i in range(0, len(texts), batch_size):
        batch_end = min(i + batch_size, len(texts))
        batch_texts = texts[i:batch_end]
        
        # Generate embeddings for this batch
        with torch.no_grad():
            batch_embeddings = model.encode(
                batch_texts, 
                prompt=task_name,
                show_progress_bar=False,
                convert_to_numpy=True
            )
        
        all_embeddings.append(batch_embeddings)
        
        # Memory management and progress
        if torch.cuda.is_available() and i % (batch_size * 5) == 0:
            torch.cuda.empty_cache()
            current_memory = torch.cuda.memory_allocated() / 1e9
            if i > 0:
                elapsed = time.time() - start_time
                rate = (i + len(batch_texts)) / elapsed
                eta = (len(texts) - i - len(batch_texts)) / rate / 60
                print(f"   📈 Progreso: {i + len(batch_texts)}/{len(texts)} ({current_memory:.2f} GB GPU, ETA: {eta:.1f}min)")
    
    # Combine all embeddings
    embeddings = np.vstack(all_embeddings)
    processing_time = time.time() - start_time
    
    print(f"✅ EmbeddingGemma finalizado!")
    print(f"⚡: {len(embeddings)/processing_time:.1f} textos/segundos")
    
    return embeddings

# Test Gemini API first
print("\n🧪 Probando conexión con Gemini API...")
try:
    test_result = genai.embed_content(
        model="models/gemini-embedding-001",
        content="Test embedding generation",
        task_type="retrieval_document"
    )
    print(f"✅ Gemini API test successful! Dimension: {len(test_result['embedding'])}")
except Exception as e:
    print(f"❌ Gemini API test failed: {e}")
    print("Please check your API key and internet connection")

# Prepare texts
texts = [prepare_text_for_embedding(pub) for pub in publications]

print(f"\n🔄 Generando embeddings para {len(texts)} publicaciones con los 2 métodos...")
print("Puede tardar varios minutos...\n")

# Generate Gemini embeddings with improved error handling
print("🌐 Método 1: Gemini API embeddings")
gemini_embeddings = generate_gemini_embeddings(texts)

# Generate EmbeddingGemma embeddings
print("\n🏠 Método 2: EmbeddingGemma embeddings")
embeddinggemma_embeddings = generate_embeddinggemma_embeddings(
    texts, 
    embeddinggemma_model, 
    task_name="STS", 
    batch_size=16
)

print(f"\n✅ Generando embeddings:")
print(f"  Gemini shape: {gemini_embeddings.shape}")
print(f"  EmbeddingGemma shape: {embeddinggemma_embeddings.shape}")

# Save embeddings for later use
embeddings_dir = Path("data/processed")
embeddings_dir.mkdir(parents=True, exist_ok=True)

np.save(embeddings_dir / "gemini_embeddings.npy", gemini_embeddings)
np.save(embeddings_dir / "embeddinggemma_embeddings.npy", embeddinggemma_embeddings)

# Save metadata
embeddings_meta = {
    'generated_at': datetime.now().isoformat(),
    'total_publications': len(publications),
    'gemini_model': "models/gemini-embedding-001",
    'embeddinggemma_model': embeddinggemma_model_id,
    'gemini_dimension': gemini_embeddings.shape[1],
    'embeddinggemma_dimension': embeddinggemma_embeddings.shape[1],
    'device_used': device,
    'processing_info': {
        'texts_processed': len(texts),
        'batch_processing': True
    }
}

with open(embeddings_dir / "embeddings_metadata.json", 'w') as f:
    json.dump(embeddings_meta, f, indent=2)

print(f"💾 Embeddings saved to: {embeddings_dir}")
```

## 4. Guardar Embeddings en ChromaDB

Vamos a crear dos colecciones de ChromaDB, una para cada método de embedding.

```python
# Initialize ChromaDB client
chroma_client = chromadb.PersistentClient(path="./chromadb")

# Create collections for both embedding methods
collection_gemini = chroma_client.get_or_create_collection(
    name="ifc_publications_gemini",
    metadata={"description": "IFC publications with Gemini API embeddings"}
)

collection_embeddinggemma = chroma_client.get_or_create_collection(
    name="ifc_publications_embeddinggemma", 
    metadata={"description": "IFC publications with EmbeddingGemma embeddings"}
)

# Prepare metadata and documents with proper None handling
documents = []
metadatas = []
ids = []

for i, pub in enumerate(publications):
    doc_id = f"pub_{i}"
    ids.append(doc_id)
    documents.append(prepare_text_for_embedding(pub))
    
    # Fix the keywords handling to properly deal with None values
    authors = pub.get('authors', [])
    if authors is None:
        authors = []
    
    keywords = pub.get('keywords', [])
    if keywords is None:
        keywords = []
    
    metadata = {
        'title': pub.get('title', '')[:500],  # Truncate if too long
        'year': str(pub.get('year', '')),
        'authors': ', '.join(authors[:3]) if authors else '',
        'keywords': ', '.join(keywords[:10]) if keywords else '',
        'doi': pub.get('doi', ''),
        'pmid': pub.get('pmid', ''),
    }
    metadatas.append(metadata)

# Add to ChromaDB collections
print("📥 Adding embeddings to ChromaDB...")

# Gemini collection
collection_gemini.add(
    embeddings=gemini_embeddings.tolist(),
    documents=documents,
    metadatas=metadatas,
    ids=ids
)

# EmbeddingGemma collection
collection_embeddinggemma.add(
    embeddings=embeddinggemma_embeddings.tolist(),
    documents=documents,
    metadatas=metadatas,
    ids=ids
)

print(f"✅ Se agregaron {len(ids)} documentos a las colecciones de ChromaDB")
print(f"  Gemini: {collection_gemini.count()} documentos")
print(f"  EmbeddingGemma: {collection_embeddinggemma.count()} documentos")
```

## 5. UMAP y Clustering

```python
# Apply UMAP to both embedding sets
print("🗺️ Aplicando reducción de dimensiones UMAP...")

umap_reducer = umap.UMAP(n_components=2, random_state=42, n_neighbors=15, min_dist=0.1)

gemini_umap = umap_reducer.fit_transform(gemini_embeddings)
embeddinggemma_umap = umap_reducer.fit_transform(embeddinggemma_embeddings)

print(f"✅ Reducción UMAP lista")
print(f"  Gemini UMAP shape: {gemini_umap.shape}")
print(f"  EmbeddingGemma UMAP shape: {embeddinggemma_umap.shape}")

# Determine optimal number of clusters using silhouette score
def find_optimal_clusters(embeddings, max_k=10):
    """Find optimal number of clusters using silhouette score"""
    silhouette_scores = []
    K_range = range(2, min(max_k + 1, len(embeddings)))
    
    for k in K_range:
        kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
        labels = kmeans.fit_predict(embeddings)
        score = silhouette_score(embeddings, labels)
        silhouette_scores.append(score)
    
    optimal_k = K_range[np.argmax(silhouette_scores)]
    return optimal_k, silhouette_scores

# Find optimal clusters for both methods
optimal_k_gemini, scores_gemini = find_optimal_clusters(gemini_umap)
optimal_k_embeddinggemma, scores_embeddinggemma = find_optimal_clusters(embeddinggemma_umap)

print(f"\n📊 Número óptimo de Clusters:")
print(f"  Gemini: {optimal_k_gemini}")
print(f"  EmbeddingGemma: {optimal_k_embeddinggemma}")

# Perform clustering with optimal k
kmeans_gemini = KMeans(n_clusters=optimal_k_gemini, random_state=42, n_init=10)
clusters_gemini = kmeans_gemini.fit_predict(gemini_umap)

kmeans_embeddinggemma = KMeans(n_clusters=optimal_k_embeddinggemma, random_state=42, n_init=10)
clusters_embeddinggemma = kmeans_embeddinggemma.fit_predict(embeddinggemma_umap)
```

## 8. Embeddings: comparar

```python
from scipy.spatial.distance import cosine
from scipy.stats import spearmanr

# Calculate pairwise cosine similarities for a sample
sample_size = min(100, len(gemini_embeddings))
sample_indices = np.random.choice(len(gemini_embeddings), sample_size, replace=False)

gemini_sample = gemini_embeddings[sample_indices]
embeddinggemma_sample = embeddinggemma_embeddings[sample_indices]

# Compute similarity matrices
def compute_similarity_matrix(embeddings):
    """Compute pairwise cosine similarity matrix"""
    n = len(embeddings)
    sim_matrix = np.zeros((n, n))
    for i in range(n):
        for j in range(i, n):
            similarity = 1 - cosine(embeddings[i], embeddings[j])
            sim_matrix[i, j] = similarity
            sim_matrix[j, i] = similarity
    return sim_matrix

print("Computing similarity matrices...")
sim_matrix_gemini = compute_similarity_matrix(gemini_sample)
sim_matrix_embeddinggemma = compute_similarity_matrix(embeddinggemma_sample)

# Flatten matrices (excluding diagonal)
mask = ~np.eye(sample_size, dtype=bool)
sim_flat_gemini = sim_matrix_gemini[mask]
sim_flat_embeddinggemma = sim_matrix_embeddinggemma[mask]

# Calculate correlation between similarity matrices
correlation, p_value = spearmanr(sim_flat_gemini, sim_flat_embeddinggemma)

print(f"\n📊 Embedding Comparison Results:")
print(f"  Sample size: {sample_size} publications")
print(f"  Spearman correlation: {correlation:.4f}")
print(f"  P-value: {p_value:.4e}")
print(f"\n  Interpretation: {'Strong' if abs(correlation) > 0.7 else 'Moderate' if abs(correlation) > 0.4 else 'Weak'} agreement between Gemini API and EmbeddingGemma")

# Visualize similarity comparison
fig = px.scatter(
    x=sim_flat_gemini,
    y=sim_flat_embeddinggemma,
    labels={'x': 'Gemini API Similarity', 'y': 'EmbeddingGemma Similarity'},
    title=f'Embedding Methods Comparison (Spearman ρ = {correlation:.3f})',
    opacity=0.3
)
fig.add_trace(go.Scatter(x=[0, 1], y=[0, 1], mode='lines', name='Perfect Agreement', line=dict(dash='dash', color='red')))
fig.update_layout(height=500, width=700)
fig.show()
```

## 12. Generate Podcast Audio with Google TTS

Si no se pueden configurar las llaves de Google tts correctamente, asegurarte de:

- Instalar dependencias 'pip install -U google-genai>=1.16.0'
- Verificar 'config.yaml'
- Configurar las variables de entorno 'export GOOGLE_TTS_API_KEY="xxx"'

```python
import os
from pathlib import Path
import wave

def read_podcast_script(script_file="./podcast_script.txt"):
    """Read podcast script from file"""
    script_path = Path(script_file)
    if not script_path.exists():
        raise FileNotFoundError(f"Script file not found: {script_file}")
    
    with open(script_path, 'r', encoding='utf-8') as f:
        content = f.read()
    
    print(f"📖 Script leído desde: {script_file}")
    print(f"📊 Longitud: {len(content)} caracteres")
    return content

def clean_script_for_tts(script):
    """Remove stage directions and formatting for TTS"""
    import re
    
    # Remove brackets and their contents (stage directions)
    cleaned = re.sub(r'\[.*?\]', '', script)
    
    # Remove "HOST:" and similar speaker labels
    cleaned = re.sub(r'^(HOST|SPEAKER|NARRATOR):\s*', '', cleaned, flags=re.MULTILINE)
    
    # Remove extra whitespace
    cleaned = re.sub(r'\n\s*\n', '\n\n', cleaned)
    cleaned = cleaned.strip()
    
    return cleaned

def save_wave_file(filename, pcm_data, channels=1, rate=24000, sample_width=2):
    """Save PCM audio data to a WAV file"""
    try:
        with wave.open(filename, "wb") as wf:
            wf.setnchannels(channels)
            wf.setsampwidth(sample_width)
            wf.setframerate(rate)
            wf.writeframes(pcm_data)
        print(f'✅ Audio guardado en {filename}')
        return True
    except Exception as e:
        print(f'❌ Error guardando audio: {e}')
        return False

def generate_audio_from_script(script_file="./podcast_script.txt", output_file="./podcast_episode.wav"):
    """
    Generate audio from existing script file using Gemini TTS
    """
    print("🎙️ GENERADOR DE AUDIO INDEPENDIENTE")
    print("=" * 50)
    
    # 1. Configurar API Key
    try:
        # Intentar cargar desde config
        from utils.config import load_config
        config = load_config()
        tts_api_key = config['api_keys'].get('google_tts', '') or config['api_keys'].get('google', '')
    except:
        # Fallback a variables de entorno
        tts_api_key = None
    
    if not tts_api_key:
        tts_api_key = os.getenv('GOOGLE_TTS_API_KEY') or os.getenv('GEMINI_API_KEY')
    
    if not tts_api_key:
        print("❌ No se encontró clave API. Configura:")
        print("   export GOOGLE_TTS_API_KEY='tu_clave_aqui'")
        print("   export GEMINI_API_KEY='tu_clave_aqui'")
        return None
    
    print(f"🔑 TTS API Key: ✅ Encontrada")
    
    # 2. Leer script del archivo
    try:
        script_content = read_podcast_script(script_file)
    except FileNotFoundError as e:
        print(f"❌ {e}")
        print(f"💡 Asegúrate de que existe el archivo: {script_file}")
        return None
    
    # 3. Limpiar y preparar texto
    clean_text = clean_script_for_tts(script_content)
    
    # Limitar longitud para TTS
    max_chars = 5000  # Límite conservador para Gemini TTS
    if len(clean_text) > max_chars:
        print(f"⚠️ Script muy largo ({len(clean_text)} chars). Recortando...")
        # Intentar cortar en una oración completa
        truncated = clean_text[:max_chars]
        last_period = truncated.rfind('.')
        if last_period > max_chars * 0.8:
            clean_text = truncated[:last_period + 1]
        else:
            clean_text = truncated
        print(f"📝 Script recortado a: {len(clean_text)} caracteres")
    
    # 4. Generar audio usando Gemini TTS
    try:
        from google import genai
        from google.genai import types
        
        # Crear cliente TTS
        tts_client = genai.Client(api_key=tts_api_key)
        
        print("🎵 Generando audio con Gemini TTS...")
        print(f"📊 Procesando: {len(clean_text)} caracteres")
        
        # Configurar prompt para español mexicano profesional
        tts_prompt = f"""Lee este guión de podcast en español mexicano con voz profesional y clara. 
        Usa un ritmo natural, pausas apropiadas, y entonación profesional:
        
        {clean_text}"""
        
        # Generar audio
        response = tts_client.models.generate_content(
            model="gemini-2.5-flash-preview-tts",
            contents=tts_prompt,
            config={"response_modalities": ['Audio']},
        )
        
        # Extraer datos de audio
        audio_blob = response.candidates[0].content.parts[0].inline_data
        
        # Guardar archivo
        output_path = Path(output_file)
        if save_wave_file(str(output_path), audio_blob.data):
            file_size = output_path.stat().st_size
            duration_est = file_size / (24000 * 2)  # Estimación basada en 24kHz, 16-bit
            
            print(f"\n🎉 ¡AUDIO GENERADO EXITOSAMENTE!")
            print(f"📁 Guardado en: {output_path}")
            print(f"📊 Tamaño: {file_size:,} bytes ({file_size/1024/1024:.1f} MB)")
            print(f"⏱️ Duración estimada: ~{duration_est:.1f} segundos ({duration_est/60:.1f} minutos)")
            print(f"🎵 Tipo MIME: {audio_blob.mime_type}")
            
            return str(output_path)
        else:
            return None
            
    except ImportError as e:
        print(f"❌ Error importando google.genai: {e}")
        print("💡 Instala: pip install -U google-genai>=1.16.0")
        return None
    except Exception as e:
        print(f"❌ Error generando TTS: {e}")
        import traceback
        traceback.print_exc()
        return None

# EJECUTAR GENERACIÓN DE AUDIO
# =================================

# Verificar que existe el script
script_file = "./podcast_script.txt"
if Path(script_file).exists():
    print(f"📄 Script encontrado: {script_file}")
    
    # Generar audio
    audio_file = generate_audio_from_script(
        script_file="./podcast_script.txt",
        output_file="./podcast_episode.wav"
    )
    
    if audio_file:
        # Mostrar reproductor de audio en Jupyter
        try:
            from IPython.display import Audio, display
            print(f"\n🎧 REPRODUCTOR DE AUDIO:")
            display(Audio(filename=str(audio_file), autoplay=False))
        except ImportError:
            print(f"💡 Para reproducir en Jupyter: Audio('{audio_file}')")
    else:
        print("❌ Falló la generación de audio")
        
else:
    print(f"❌ No se encontró el archivo: {script_file}")
    print("💡 Primero ejecuta la celda de generación de script o crea el archivo manualmente")
```

```python
# Optional: Use Gemini to generate human-readable cluster descriptions
def generate_cluster_descriptions(cluster_data, num_samples=5):
    """Generate natural language descriptions of clusters using Gemini"""
    
    descriptions = {}
    model = genai.GenerativeModel('gemini-1.5-flash')
    
    for cluster_id in sorted(cluster_data['Cluster'].unique()):
        cluster_pubs = cluster_data[cluster_data['Cluster'] == cluster_id]
        
        # Get sample titles and keywords
        sample_titles = cluster_pubs['Title'].head(num_samples).tolist()
        sample_keywords = cluster_pubs['Keywords'].head(num_samples).tolist()
        
        prompt = f"""Based on these research publication titles and keywords from a cluster, provide a brief 2-3 sentence description of the main research theme:

Titles:
{chr(10).join(f'{i+1}. {t}' for i, t in enumerate(sample_titles))}

Keywords:
{chr(10).join(sample_keywords)}

Provide a concise, scientific description of this research cluster's focus:"""
        
        response = model.generate_content(prompt)
        descriptions[cluster_id] = response.text.strip()
    
    return descriptions

# Generate descriptions for Gemini clusters
print("📝 Generating cluster descriptions with AI...\n")
cluster_descriptions = generate_cluster_descriptions(df_gemini)

for cluster_id, description in cluster_descriptions.items():
    print(f"Cluster {cluster_id}:")
    print(f"  {description}\n")
```
